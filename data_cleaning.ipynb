```python
### Data Types ###

data_name.info()    # check types of the columns

data_name['column_name'] = data_name['column_name'].astype('int')    # convert a column to a different data type

# str - string (text) data type
# int - integer (whole number) data type
# float - floating point (decimal number) data type
# bool - boolean (True/False) data type
# dict - dictionary (key-value pairs) data type
# list - list (ordered collection) data type


# remove and replace characters (ex. commas as considered object if number has colum in dataframe)
data_name['column_name'] = data_name['column_name'].str.replace('remove', 'replace with')       # ex. (',', '') - remove , and replace with nothing ('')


## Date & Time

# specify date and time data to pandas
data_name = pd.read_csv('data_name.csv', parse_dates = ['date_column_name'])
                        
# after importing
data_name['date_column_name'] = pd.to_datetime(data_name['date_column_name'])

# if date elements stored separately can combine
data_name['date_column_name'] = pd.to_datetime(data_name[['day', 'month', 'year']])

# select dat, month, or year from the column (.dt.day, .dt.month, .dt.year)
data_name['new_date_column_name'] = data_name['date_column_name'].dt.month
```

```python
### Missing Data ###

## options
# drop missing values (usually if only 5%)
# imputation - replace with mean, median, mode (for mode specify .mode()[0], to pick first one as may be multiple)
# imputation by sub-group


## Drop

five_percent = len(data_name) * 0.05                                    # given the length of the data, calculate what 5% would be

(data_name.isna().sum())                                                # count number of missing values

dropping = data_name.columns[data_name.isna().sum() <= five_percent]    # determine which ones to drop (with na)

data_name.dropna(subset = dropping, inplace = True)                     # drop na usign subset to drop no more than 5%, inplace to update set

print(data_name.isna().sum())                                           # check na again


## Imputation

columns_missing = data_name.columns[data_name.isna().sum() > 0]         # get columns that have na values

for col in columns_missing[ : -1]: 
    data_name[col] = data_name[col].fillna(data_name[col].mean())                     # [ : -1] to get all, .fillna(), get the mean for each column


## Imputation by sup-group

data_name_dict = data_name.groupby('column_of_interest')['column_name'].mean().to_dict()    # store grouped mean data as a dictionary

data_name['column_name'] = data_name['column_name'].fillna(data_name['sub_group_column'].map(data_name_dict))

```
