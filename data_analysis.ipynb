```python
### Probability ###

np.random.seed(5472)                        # set seed for reproducibility

data_name.sample(n = 10)                    # randomly sample 10 rows from a dataframe (can specify any number of rows to sample)

data_name.sample(n = 10, replace = True)    # randomly sample 10 rows with replacement



## Distributions

# Discrete
sample_name = data_name.sample(n = 1000, replace = True)

sample_name['column_name'].hist(bins = 30)   
plt.show()


# Uniform
from matplotlib.pylab import average
from scipy.stats import uniform

uniform.rvs(lower limit, upper limit, size = 1000)    # generate 1000 random variates from a uniform distribution defined by the specified limits

uniform.cdf(value of interest, lower limmit, upper limit)   # gives the provability of a value being less than or equal to the value of interest in a uniform distribution defined by the specified limits


# Binomial
from scipy.stats import binom

binom.rvs(n = number of trials, p = probability of success, size = 1000)                # generate 1000 random variates from a binomial distribution defined by the specified parameters

binom.pmf(k = number of successes, n = number of trials, p = probability of success)    # gives the probability of observing exactly k successes in a binomial distribution defined by the specified parameters

binom.cdf(k = number of successes, n = number of trials, p = probability of success)    # gives the probability of observing k or fewer successes in a binomial distribution defined by the specified parameters

# expected value = n * p


# Normal
from scipy.stats import norm

norm.rvs(mean, standard deviation, size = 1000)          # generate 1000 random variates from a normal distribution defined by the specified mean and standard deviation

norm.cdf(value of interest, mean, standard deviation)    # gives the probability of a value being less than or equal to the value of interest in a normal distribution defined by the specified mean and standard deviation

norm.ppf(percentile, mean, standard deviation)           # gives the probability density of a value at a specific percentile (prob of being below or at that percentile) in a normal distribution defined by the specified mean and standard deviation

# Central Limit Theorem - distribution of sample means, sd, proportion will appraoch normal distribution as sample size increases, regardless of the shape of the population distribution


# Poisson

# poisson process - events happen at rando but at a certain rate 
# poisson distribution - probability of a given number of events happening in a fixed interval of time
# discrete (represents count of events)
# lambda (λ) = the average number of events per interval

from scipy.stats import poisson

poisson.rvs(mu = average number of events, size = 1000)             # generate 1000 random variates from a Poisson distribution defined by the specified average number of events

poisson.pmf(k = number of events, mu = average number of events)    # gives the probability of observing exactly k events in a Poisson distribution defined by the specified average number of events

poisson.cdf(k = number of events, mu = average number of events)    # gives the probability of observing k or fewer events in a Poisson distribution defined by the specified average number of events


# Exponential

# probability distribution of the time between events in a Poisson process (ex. propability of 7 months between events if the average time between events is 5 months)
# continuous (represents time)
# lambda (λ) = rate, interested in 1/lambda (λ) = mean time between events

form scipy.stats import expon

expon.rvs(scale = mean time between events, size = 1000)    # generate 1000 random variates from an exponential distribution defined by the specified mean time between events

expon.cdf(value of interest, scale = mean time between events)    # gives the probability of a value being less than or equal to the value of interest in an exponential distribution defined by the specified mean time between events


# t-distribution

# degrees of freedom (df) = number of independent pieces of information that went into the estimate of the variance (n - 1 for a single sample)
# as df increases, t-distribution approaches normal distribution

from scipy.stats import t

t.rvs(df = degrees of freedom, size = 1000)    # generate 1000 random variates from a t-distribution defined by the specified degrees of freedom

t.cdf(value of interest, df = degrees of freedom)    # gives the probability of a value being less than or equal to the value of interest in a t-distribution defined by the specified degrees of freedom
```
